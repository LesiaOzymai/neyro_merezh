```python

import os
import re
import math
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm

import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, precision_score, recall_score, f1_score

import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
STOPWORDS = set(stopwords.words('english'))

# GPU check 
gpus = tf.config.list_physical_devices('GPU')
print("GPUs found:", gpus)
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Could not set memory growth:", e)

# text preprocessing 
def clean_text(text, remove_stopwords=True):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    # remove URLs, emails
    text = re.sub(r'http\S+|www\S+|https\S+', ' ', text)
    text = re.sub(r'\S+@\S+', ' ', text)
    # remove non-alphanumeric
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    # collapse spaces
    text = re.sub(r'\s+', ' ', text).strip()
    if remove_stopwords:
        tokens = [w for w in text.split() if w not in STOPWORDS]
        return " ".join(tokens)
    return text

# Load dataset
def load_yelp_from_kaggle(path='/kaggle/input/yelp-review-polarity'):
    
    candidates = [
        os.path.join(path, 'yelp_review_polarity_csv'),
        path
    ]
    for base in candidates:
        if not os.path.exists(base):
            continue
        
        for fname in ('train.csv', 'test.csv', 'train.csv.gz', 'test.csv.gz'):
            p = os.path.join(base, fname)
            if os.path.exists(p):
                print("Loading from", p)
                df_train = pd.read_csv(p, header=None, names=['label','text'], encoding='utf-8', sep=',', quotechar='"', engine='python')
                
                return df_train, None
        
        csvs = [os.path.join(base, f) for f in os.listdir(base) if f.endswith('.csv')]
        if csvs:
            print("Found CSVs:", csvs[:5])
            
            train = None; test = None
            for c in csvs:
                if 'train' in c.lower():
                    train = c
                if 'test' in c.lower():
                    test = c
            if train:
                df_train = pd.read_csv(train, header=None, names=['label','text'], encoding='utf-8', sep=',', quotechar='"', engine='python')
                df_test = pd.read_csv(test, header=None, names=['label','text'], encoding='utf-8', sep=',', quotechar='"', engine='python') if test else None
                return df_train, df_test
    return None, None

def load_yelp_with_tfds():
    try:
        import tensorflow_datasets as tfds
        print("Trying to load via tensorflow_datasets (may download)...")
        
        data, info = tfds.load('yelp_polarity_reviews', split=['train', 'test'], with_info=True, as_supervised=True)
        train_ds, test_ds = data
        
        def tfds_to_df(ds):
            texts=[]; labels=[]
            for x,y in tfds.as_numpy(ds):
                texts.append(x.decode('utf-8') if isinstance(x, bytes) else x)
                labels.append(int(y))
            return pd.DataFrame({'label': labels, 'text': texts})
        return tfds_to_df(train_ds), tfds_to_df(test_ds)
    except Exception as e:
        print("tfds load failed:", e)
        return None, None


train_df, test_df = load_yelp_from_kaggle()
if train_df is None:
    train_df, test_df = load_yelp_with_tfds()

if train_df is None:
    raise RuntimeError("Could not find Yelp dataset locally or via tfds. On Kaggle add dataset 'Yelp Review Polarity' or enable internet for tfds.")

print("Raw train rows:", len(train_df), "Raw test rows:", len(test_df) if test_df is not None else 0)

def normalize_labels(df):
    if df is None: 
        return None
    # if labels are 1 or 2 -> convert to 0/1
    unique = set(df['label'].astype(int).unique().tolist())
    if unique.issubset({1,2}):
        df['label'] = df['label'].astype(int).apply(lambda x: 0 if x==1 else 1)
    else:
        df['label'] = df['label'].astype(int)
    return df

train_df = normalize_labels(train_df)
test_df = normalize_labels(test_df)

# If only train available, split it
if test_df is None or len(test_df)==0:
    print("No test set found — splitting training set into train+test (80/20).")
    train_df, test_df = train_test_split(train_df.sample(frac=1, random_state=42), test_size=0.2, stratify=train_df['label'], random_state=42)

from tqdm.auto import tqdm
tqdm.pandas()
# Preprocess text 
print("Preprocessing texts...")
train_df['clean_text'] = train_df['text'].progress_apply(lambda x: clean_text(x, remove_stopwords=True))
test_df['clean_text']  = test_df['text'].progress_apply(lambda x: clean_text(x, remove_stopwords=True))

# Subsample for quicker experiments 
USE_SUBSAMPLE = False
SUBSAMPLE_SIZE = 50000  
if USE_SUBSAMPLE:
    train_df = train_df.sample(n=min(SUBSAMPLE_SIZE, len(train_df)), random_state=42).reset_index(drop=True)
    print("Subsampled train to", len(train_df))

# Tokenization + sequences
MAX_NUM_WORDS = 20000   # vocabulary size
MAX_SEQ_LEN = 100       # sequence length (pad/truncate)

tokenizer = Tokenizer(num_words=MAX_NUM_WORDS, oov_token='<OOV>')
tokenizer.fit_on_texts(train_df['clean_text'].values)
X_train_seq = tokenizer.texts_to_sequences(train_df['clean_text'].values)
X_test_seq  = tokenizer.texts_to_sequences(test_df['clean_text'].values)

X_train = pad_sequences(X_train_seq, maxlen=MAX_SEQ_LEN, padding='post', truncating='post')
X_test  = pad_sequences(X_test_seq,  maxlen=MAX_SEQ_LEN, padding='post', truncating='post')
y_train = train_df['label'].values.astype('int32')
y_test  = test_df['label'].values.astype('int32')

print("Shapes:", X_train.shape, y_train.shape, X_test.shape, y_test.shape)
print("Vocab size (tokenizer):", min(MAX_NUM_WORDS, len(tokenizer.word_index)+1))

# Build model
EMBED_DIM = 128
LSTM_UNITS = 128
DROPOUT = 0.4

model = Sequential([
    Embedding(input_dim=min(MAX_NUM_WORDS, len(tokenizer.word_index)+1),
              output_dim=EMBED_DIM,
              input_length=MAX_SEQ_LEN),
    Bidirectional(LSTM(LSTM_UNITS, return_sequences=False, dropout=0.2, recurrent_dropout=0.2)),
    Dropout(DROPOUT),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()

# Callbacks
model_dir = '/kaggle/working/lstm_yelp'
os.makedirs(model_dir, exist_ok=True)
checkpoint_path = os.path.join(model_dir, 'best_model.h5')

callbacks = [
    EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True, verbose=1),
    ModelCheckpoint(checkpoint_path, monitor='val_loss', save_best_only=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1)
]

# Train
EPOCHS = 6
BATCH_SIZE = 256

history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=callbacks,
    verbose=1
)

# Save final model
final_model_path = os.path.join(model_dir, 'final_lstm_model.h5')
model.save(final_model_path)
print("Saved model to", final_model_path)

# Evaluation
print("Evaluating on test set...")
loss, acc = model.evaluate(X_test, y_test, verbose=1)
print(f"Test Loss: {loss:.4f}, Test Accuracy: {acc:.4f}")

# Predictions
y_pred_prob = model.predict(X_test, batch_size=512, verbose=1).ravel()
y_pred = (y_pred_prob >= 0.5).astype(int)

print("Classification report (test):")
print(classification_report(y_test, y_pred, digits=4))

prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1  = f1_score(y_test, y_pred)
print(f"Precision: {prec:.4f}, Recall: {rec:.4f}, F1: {f1:.4f}")

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion matrix:\n", cm)

# Plot training history
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(history.history.get('loss', []), label='train_loss')
plt.plot(history.history.get('val_loss', []), label='val_loss')
plt.title('Loss')
plt.legend()
plt.grid(True)

plt.subplot(1,2,2)
plt.plot(history.history.get('accuracy', []), label='train_acc')
plt.plot(history.history.get('val_accuracy', []), label='val_acc')
plt.title('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

# Confusion matrix heatmap 
try:
    import seaborn as sns
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['neg','pos'], yticklabels=['neg','pos'])
    plt.xlabel('Predicted'); plt.ylabel('True'); plt.title('Confusion Matrix')
    plt.show()
except Exception as e:
    print("seaborn not available, skipping heatmap:", e)

# Sentiment analysis helper
def analyze_sentiment(text, model=model, tokenizer=tokenizer, seq_len=MAX_SEQ_LEN):
    cleaned = clean_text(text, remove_stopwords=True)
    seq = tokenizer.texts_to_sequences([cleaned])
    padded = pad_sequences(seq, maxlen=seq_len, padding='post', truncating='post')
    prob = model.predict(padded)[0][0]
    label = "Positive" if prob >= 0.5 else "Negative"
    return {'label': label, 'probability': float(prob)}


samples = [
    "Absolutely loved the atmosphere and the food was fantastic!",
    "I had to wait forever and the service was terrible.",
    "It was okay, nothing special but not bad.",
    "Worst experience ever. I will never go back."
]
for s in samples:
    print(s, "->", analyze_sentiment(s))

# Save tokenizer for reuse
import pickle
tk_path = os.path.join(model_dir, 'tokenizer.pkl')
with open(tk_path, 'wb') as f:
    pickle.dump(tokenizer, f)
print("Tokenizer saved to", tk_path)


print("Finished. Model and tokenizer saved in", model_dir)
```
